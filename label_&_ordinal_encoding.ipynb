# **ORDERED ENCODING**
In this the encoding is applied depend on the ordered


```python
'''
# EXAMPLE

In KitchenQual there are 4 classes

Ex == Excellent
Gd == Good
Ta == Typical/Average
Fa == Fair
'''
'''
We can gieve default priority number to each class like Ex=1, Gd=2 etc
```

```python
df['KitchenQual'].value_counts()
```

```python
# we use dictionary to give priority

order_label = {'Ex':4,'Gd':3,'TA':2,'Fa':1}
```

```python
df2['Kitchen_ord_enc']=df['KitchenQual'].map(order_label)
df2
```
